```python
import numpy as np

def mean_std(ax, energies, title, max_energy=665., scale_factor=1.0, RT=False):
    # Filter energies within the relevant range
    cutoff_energies = energies[(energies >= 550.) & (energies <= 662)] 
    
    # Define histogram bins
    bin_num = int((662 - 550) * 4)
    counts, bin_edges = np.histogram(cutoff_energies, bins=bin_num, range=(550, 662))
    
    # Determine peak count and threshold
    peak_count = np.max(counts)
    threshold = 0.15 * peak_count

    # Find min_energy
    min_energy = 550
    consecutive_count = 0  
    count_num = 3  # Number of consecutive bins to confirm threshold

    for i in range(bin_num - 1, -1, -1):
        if counts[i] <= threshold:
            consecutive_count += 1
        else:
            consecutive_count = 0  
        
        if consecutive_count >= count_num:
            min_energy = bin_edges[i - count_num - 1]  
            break
    
    if RT:
        min_energy = 475
    
    # Filter energies based on determined range
    energies = energies[(energies >= min_energy) & (energies <= max_energy)]  
    
    # Compute statistics
    mean = np.mean(energies)
    std_dev = np.std(energies)
    FWHM = 2.355 * std_dev
    
    # Compute statistical uncertainty
    N = len(energies)
    if N > 1:
        FWHM_uncertainty = 2.355 * (std_dev / np.sqrt(2 * N))
    else:
        FWHM_uncertainty = np.nan  # Avoid division by zero
    
    # Plot histogram
    bin_num = int((max_energy - min_energy) * 4)
    ax.hist(energies, bins=bin_num, range=(min_energy, max_energy), alpha=0.6, color='g')#,label=f'FWHM$_{{eff}}$ = {FWHM:.2f} ± {FWHM_uncertainty:.3f} keV')
    
    # Plot vertical lines
    ax.axvline(mean, color='r', linestyle='dashed', linewidth=4, label=f'$μ$ = {mean:.2f} keV')
    ax.axvline(mean - std_dev, color='b', linestyle='dashed', linewidth=4, label=f'$σ$ = {std_dev:.2f} keV')
    ax.axvline(mean + std_dev, color='b', linestyle='dashed', linewidth=4)
    
    # Set labels and title
    font_size = 12 * scale_factor
    ax.set_xlim(min_energy, max_energy)
    ax.set_xlabel('Energy (keV)', fontsize=font_size*1.2)
    ax.set_ylabel('Counts', fontsize=font_size*1.2)
    ax.set_title(title, fontsize=1.4 * font_size)
    ax.legend(fontsize=1.6*font_size)
    ax.tick_params(labelsize=font_size*1.2)
    
    return FWHM, FWHM_uncertainty
```

Examining annealing at 80C, we try to separate primary and secondary defects. We assume, based on observations from literature, that the annealing at 80C must have a larger time constant than annealing at 100C. Approximating this time constant, we attempt to fit the time constant for secondary defects, assuming amplitudes for primary decay that corresponds to the the FWHM after 80C anneals, and for the secondary decay that corresponds to the FWHM after RT anneal

First we'll try the scipy curve fit method. We onlyhave three datapoints at 80C :( so we need to fix one of them in order to use curve fit. I fix the time constant of the primary decay ($tau_1$) to 5X the time constant we found for primary defects at 100C (~5 x 75 = 375). This is based on work in Brown and Fletcher 1958. Note that we set an upper bound on primary defect amplitude at 38 keV FWHM since this is the post radiation resolution aka for primary defects and we don't unergo any processes afterward which would produce more primary defects. The amplitude for secondary defects is unbounded as Fourcher 1991 observes that RT annealing can tranform primary defects into secondary defects. 

```python
def double_exp_fixed_tau1(t, A1, A2, tau2, C):
    tau1 = 118.0  # Fix tau1 to a known value (e.g., 1.0)
    return A1 * np.exp(-t / tau1) + A2 * np.exp(-t / tau2) + C

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

def fit_double_exponential_fixed_tau1(t_data, y_data, y_err, initial_guess):
    
    tau1_fixed = 118.0  # Fixed tau1 value

    bounds = ([0, 0, 0, 0], [np.inf, np.inf, np.inf, 2.5])
    
    t_array = np.array(t_data)
    y_array = np.array(y_data)
    y_err_array = np.array(y_err)
    
    # Fit the data using curve_fit, where only A1, A2, and tau2 are fitted
    popt, pcov = curve_fit(double_exp_fixed_tau1, t_array, y_array, sigma=y_err, absolute_sigma=True, p0=initial_guess, bounds=bounds)
    
    # Extract the optimal parameters
    A1_fit, A2_fit, tau2_fit, C_fit = popt
    print(f"Fitted Parameters: A1 = {A1_fit}, A2 = {A2_fit}, tau2 = {tau2_fit},  C = {C_fit}, tau1 (fixed) = {tau1_fixed}")
    
    # Calculate the standard deviations (errors) on the parameters
    perr = np.sqrt(np.diag(pcov))  # Standard deviations (errors)
    print("Parameter errors:", perr)
    
    # Generate fitted data for the whole curve
    t_fit = np.linspace(min(t_data), max(t_data), 1000)
    y_fit = double_exp_fixed_tau1(t_fit, *popt)
    
    # Generate the individual exponential components
    exp1 = A1_fit * np.exp(-t_fit / tau1_fixed) +C_fit  # First exponential component
    exp2 = A2_fit * np.exp(-t_fit / tau2_fit)   # Second exponential component
    
    # Compute residuals
    y_fit_data = double_exp_fixed_tau1(t_array, *popt)
    residuals = y_data - y_fit_data

    # Compute chi-squared
    chi_sq = np.sum((residuals / y_err) ** 2)
    reduced_chi_sq = chi_sq / (len(t_data) - len(popt))  # Reduced χ²

    print(f'Reduced chi-squared: {reduced_chi_sq}')
    
    # Plot the original data points
    plt.figure(figsize=(8, 6))
    plt.scatter(t_data, y_data, color='red', label='Data points')
    
    # Plot the full fitted curve
    plt.plot(t_fit, y_fit, label='Fitted curve', color='blue', linewidth=2)
    
    # Plot the individual exponential components
    plt.plot(t_fit, exp1, '--', label=f'Component 1: A1 * exp(-t/tau1) + C', color='green')
    plt.plot(t_fit, exp2, '--', label=f'Component 2: A2 * exp(-t/tau2) + C', color='orange')
    
    # Labels and legend
    plt.xlabel('Time (t)')
    plt.ylabel('Value (y)')
    plt.legend()
    plt.title('Double Exponential Fit with Individual Components')
    plt.grid(True)
    plt.show()

    return popt, reduced_chi_sq


    
guess=[25, 5, 5, 1]
errors=[ 0.75*x for x in total_error] 
print(errors)
fit_double_exponential_fixed_tau1(times, energies100C, errors,  guess)
print(energy1)
```
